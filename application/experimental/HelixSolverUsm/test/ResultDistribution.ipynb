```python
import matplotlib.pyplot as plt
import math
```

```python
indexes = []
numHits = []
rs = []
phis = []

path = '../../../../log.txt'
with open(path, 'r') as file:
    for line in file:
        if 'DEBUG' not in line:
            continue

        line = line[line.find('DEBUG') + 6:]
        line = line.split(',')
        if len(line) < 4:
            continue

        line = [x.strip() for x in line]
        indexes.append(int(line[0]))
        numHits.append(int(line[1]))
        rs.append(float(line[2]))
        phis.append(float(line[3]))

for i in range(10):
    print(indexes[i], numHits[i], rs[i], phis[i])
```

```python
qOverPts = [1 / (r * 2) for r in rs]
```

```python
figure_size = (10, 8)
plt.figure(figsize=figure_size)
ax = plt.axes()
ax.set_facecolor('black')
ax.set_xlabel('qOverPt')
ax.set_ylabel('phi')
ax.grid(True, alpha=0.3, color='white')

scatter = ax.scatter(qOverPts, phis, alpha=0.1, c='red', s=1)

plt.tight_layout()
plt.title('qOverPt vs phi')
plt.show()
```
